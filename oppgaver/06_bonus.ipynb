# 06 – Bonus: ditt eget prosjekt

Velg et datasett du er nysgjerrig på, og bygg din egen pipeline fra
rådata til trening, evaluering og tolkning av resultater.

## Finn noe du vil undersøke

Se [datasettene i torchvision](https://docs.pytorch.org/vision/stable/datasets.html) for en samling med strukturerte datasett, eller finn noe helt annet!

Feks [EuroSAT - bilder fra sentinel 2](https://docs.pytorch.org/vision/stable/generated/torchvision.datasets.EuroSAT.html#torchvision.datasets.EuroSAT)

Velg et datasett du har lyst til å jobbe med

- Hva inneholder dataene, og hva vil du at modellen skal forutsi?
- Hvordan lastes de ned? Sjekk størrelse, eventuelle ekstra avhengigheter og
  om nedlasting må gjøres manuelt. Lagre data i `data/` i prosjektroten.
- Hvordan er trenings- og testdata delt? Hva med klassene?


## Lag en pipeline

- Definer datasett, dataloader, modell, etc
- Tren modellen og visualiser resultatene


## Fortell hva du fant

Oppsummer prosjektet slik at sidemannen kan forstå hva du har gjort, og hvordan eksperimentet kan gjenskapes:

- Hva prøvde du å finne ut, og hvordan bygde du pipelinen?
- Hvilke datautvalg og innstillinger brukte du? Hva viste målingene og figurene?
- Hvordan gikk forsøket sammenlignet med forventningene?
- Hvilke feil eller begrensninger fant du? Hva ville du undersøkt videre?
