## Eksempel: matematikk med NumPy

En Python-liste er en samling objekter. En NumPy-tabell lar oss regne på alle
verdiene samtidig. `*` betyr elementvis multiplikasjon.
`shape` beskriver størrelsen langs hver akse; `ndim` teller aksene.
Vi regner med et lite sett målinger: radene er dager, kolonnene er målepunkter.


```python
maalinger = np.array([[2.0, 4.0, 6.0], [3.0, 5.0, 7.0]])
print("Form:", maalinger.shape, "Dimensjoner:", maalinger.ndim)
print("Doblet:", maalinger * 2)
print("Første rad:", maalinger[0])
print("Gjennomsnitt per rad:", maalinger.mean(axis=1))
```

## Eksempel: fra NumPy til PyTorch

En tensor ligner en NumPy-tabell. PyTorch kan i tillegg følge beregninger og
beregne gradienter. Det kommer vi tilbake til. Her bruker vi CPU og `float32`,
en datatype for desimaltall som også brukes av modellene våre.


```python
tabell = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.float32)
tensor = torch.tensor(tabell)
print(tensor)
print("Form:", tensor.shape, "Datatype:", tensor.dtype)
print("Antall dimensjoner:", tensor.ndim, "Antall elementer:", tensor.numel())
print("Gjennomsnitt per rad:", tensor.mean(dim=1))
```

## Eksempel: endre form uten å endre verdier

`reshape` fordeler de samme elementene over nye dimensjoner.
`-1` betyr «regn ut denne størrelsen». Antall elementer må være uendret.
`unsqueeze` legger til en akse med størrelse 1; `squeeze` fjerner slike akser.


```python
verdier = torch.arange(12, dtype=torch.float32)
rektangel = verdier.reshape(3, 4)
print(rektangel)
print("Flat:", rektangel.reshape(-1).shape)
print("Med ekstra akse:", rektangel.unsqueeze(0).shape)
```
