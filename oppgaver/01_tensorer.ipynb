## Eksempel: en tabell med tall

En **skalar** er et tall, en **vektor** er en liste og en **matrise** er en
tabell. Tensor er fellesnavnet, også når det finnes flere akser.
`shape` gir størrelsen langs aksene; `ndim` teller aksene og `numel()` teller
alle tallene. Vi bruker `float32` til desimaltall og `int64` til klasseindekser.
Indekser starter på 0. I en matrise velger `x[0]` første rad, `x[:, 0]` første kolonne.


```python
measurements = torch.tensor([[2.0, 4.0, 6.0], [3.0, 5.0, 7.0]])
print(measurements)
print("Form:", measurements.shape, "Akser:", measurements.ndim, "Elementer:", measurements.numel())
print("Første rad:", measurements[0], "Første kolonne:", measurements[:, 0])
print("Datatype:", measurements.dtype)
```

## Eksempel: summer langs en akse

`dim` sier hvilken akse vi regner over og fjerner. `dim=1` summerer kolonnene
innen hver rad. `dim=0` summerer radene innen hver kolonne. `keepdim=True`
beholder den reduserte aksen med størrelse 1, noe vi snart trenger.


**Broadcasting:** Ved regning mellom tensorer sammenlignes formene fra høyre. Størrelsene må være like, eller én av dem må være 1. Da kan `[2, 1]` utvides til `[2, 3]`, mens `[2]` ikke passer med `[2, 3]`.


```python
print("Sum per rad:", measurements.sum(dim=1))
print("Sum per kolonne:", measurements.sum(dim=0))
print("Med beholdt akse:", measurements.mean(dim=1, keepdim=True).shape)
```

## Eksempel: elementvis gange og matrisemultiplikasjon

`*` ganger tall på tilsvarende plasser. `@` lager vektede summer:
`[B, D] @ [D, K] → [B, K]`. Ett resultattall er én rad ganget med én kolonne,
summert. De indre størrelsene må være like. Regn første resultattall for hånd.

`torch.dot(x[0], w[:, 0])` beregner prikkproduktet mellom første rad og første
kolonne: gang sammen tilsvarende verdier og summer, her `1*2 + 2*1 = 4`.


```python
x = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
w = torch.tensor([[2.0, 0.0], [1.0, 3.0]])
print("Elementvis:\n", x * w)
print("Matriseprodukt:\n", x @ w)
print("Første tall i produktet:\n", torch.dot(x[0], w[:, 0]).item())
```

## Bonus: NumPy

NumPy brukes ofte til numerisk matematikk og tallberegninger utenfor PyTorch. Mye av tankegangen er
lik: `np.array`, `.shape`, `.reshape` og `@`. Reduksjoner bruker `axis` i stedet
for `dim`.


```python
import numpy as np

numpy_array = np.array([[1.0, 2.0], [3.0, 4.0]], dtype=np.float32)
tensor = torch.tensor(numpy_array)
print(numpy_array.mean(axis=1), tensor.mean(dim=1))
```
