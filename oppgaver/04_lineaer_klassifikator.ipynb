## Eksempel: én oppdatering på Fashion-MNIST

Dersom du har brukt `model.fit(X, y)` i scikit-learn, har biblioteket gjort
arbeidet med å tilpasse modellens parametere. Her skriver vi treningsløkken
selv. 

Vi bruker Adam til å tilpasse vektene og konstantleddene, altså modellens parametre.

**For hver batch utføres følgende steg:**

1. Nullstill gamle gradienter: `demo_optimizer.zero_grad()`.
2. Beregn skårer fra bildene: `scores = demo_model(images)`.
3. Beregn tapet mot riktige klasser: `loss = demo_loss_fn(scores, labels)`.
4. Beregn gradientene: `loss.backward()`.
5. Oppdater parametrene: `demo_optimizer.step()`.

Prosedyren gjentas for hver batch gjennom alle epokene.

En gradient beskriver hvordan tapet endres ved små endringer i parametrene.
`backward()` beregner gradientene automatisk.
`step()` bruker gradientene til å oppdatere parametrene i henhold til Adam.

PyTorch summerer gradienter ved hvert kall til `backward()`. Vi nullstiller
dem før neste batch slik at oppdateringen bruker gradientene fra denne batchen.

Vi beregner tapet på **samme batch** før og etter oppdateringen for å undersøke
virkningen.


```python
demo_model = nn.Sequential(nn.Flatten(), nn.Linear(784, 10))
demo_loss_fn = nn.CrossEntropyLoss()
demo_optimizer = torch.optim.Adam(demo_model.parameters(), lr=0.001)

images, labels = next(iter(train_loader))
demo_model.train()
demo_optimizer.zero_grad()  # 1. Nullstill gamle gradienter
scores = demo_model(images)  # 2. Beregn skårer
loss = demo_loss_fn(scores, labels)  # 3. Beregn tap
loss.backward()  # 4. Beregn gradienter
demo_optimizer.step()  # 5. Oppdater parametere
print("Tap før oppdateringen:", loss.item())

# Mål tapet på de samme bildene med de oppdaterte parametrene.
with torch.no_grad():
    loss_after = demo_loss_fn(demo_model(images), labels)
print("Tap etter oppdateringen:", loss_after.item())
```
